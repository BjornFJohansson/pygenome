Researchers in the genome deletion project systematically deleted genes in *Saccharomyces cerevisiae* BY4741 and  using a **PCR-based gene deletion strategy** in the **Yeast Genome Deletion Project**. The process involved the following steps:

1. **Designed Primers & Amplified the Disruption Cassette**  
   - They designed gene-specific primers with ~45 bp flanking sequences homologous to the target gene’s upstream and downstream regions.  
   - They used these primers to amplify a **disruption cassette** containing a **selectable marker** (e.g., *kanMX*, which conferred G418 resistance).  
2. **Transformed Yeast & Replaced the Target Gene**  
   - They introduced the PCR-generated disruption cassette into yeast cells using **electroporation or lithium acetate transformation**.  
   - Homologous recombination between the cassette's flanking regions and the yeast genome **replaced the target gene** with the selectable marker.  

3. **Selected and Verified Transformants**  
   - They selected transformants on G418-containing medium.  
   - They confirmed successful gene replacement using **colony PCR or Southern blot**.  
   - In some strains, they incorporated **barcode sequences** to enable easy identification in pooled screens.  

Two PCR strategies were used. Strategy 1 implies amplifying a cassette from the pFA6a-kanMX4 plasmid using the UPTAG_primer and DNTAG_primer. These primers contain two barcode sequences, TAG1 and TAG2 as well as the constant regions U1, U2, D1 and D2. In a second PCR, Primers UPstream45 and  DNstream45 were used to amplify the cassette with tails that gave the PCR product flanking homology for integration.

## Strategy 1:

          First PCR:
          
          UPTAG_primer = U1-TAG1-U2-> 
                                    KanMX4
                                         <-D1-TAG2-D2 = DNTAG_primer
                                      
                         U1-TAG1-U2-KanMX4-D1-TAG2-D2                     Inner cassette

           Second PCR:
    
              UPstream45->
                         U1-TAG1-U2-KanMX4-D1-TAG2-D2
                                                    <-DNstream45

              UPstream45-U1-TAG1-U2-KanMX4-D1-TAG2-D2-DNstream45           Outer cassette

In The alternative strategy, Strategy 2, The DNstream45 primer was made to anneal directly to the KanMX4 cassette and no DNTAG_primer was used. This means that only on barcode was incorporated in the deletion cassette.

## Strategy 2:

           UPTAG_primer = U1-TAG1-U2-> 
                                     KanMX4
                                          <-DNstream45 = DNstream45 primer
                                         
                           U1-TAG1-U2-KanMX4-DNstream45                    Inner cassette

                UPstream45->
                           U1-TAG1-U2-KanMX4-DNstream45                    Outer cassette
                                           <-DNstream45

                UPstream45-U1-TAG1-U2-KanMX4-DNstream45

For some deletions, a UPstream90, DNstream90 primer pair were synthesized with twice the lenght of homology compared to the UPstream45 and DNstream45 primers.

The code cell below confirms that all rows contain an UPTAG primer sequence and that all UPTAG sequences contain the U1 and U2 sequences.
All dntag primer sequences contain the D1 and D2 sequences, but not all rows contain a dntag primer sequence.

```python
no_dntag = 0
for index, row in df.iloc[1:].iterrows():
    row_dict = row.to_dict()
    uptag = row_dict["UPTAG_primer_sequence"]
    dntag = row_dict["DNTAG_primer_sequence"]
    tag1 = row_dict["UPTAG_sequence_20mer"]
    assert U1 + tag1 + U2 in uptag
    if dntag:
        tag2 = row_dict["DNTAG_sequence_20mer"]
        assert D1 + tag2 + D2 in dntag        
    else:
        no_dntag += 1
```

192 of the 6363 rows (~3%) contain no DNTAG_primer_sequence which is also confirmed on the original [website](http://chemogenomics.pharmacy.ubc.ca/GGCN_Lab/SGDP/group/yeast_deletion_project/project_desc.html). 

```python
assert no_dntag == 192
```

```python
from pydna.primer import Primer
from pydna.amplify import pcr
from pydna.readers import read
```

```python
pFA6 = read("data/pFA6a-kanMX4.gb")
```

```python
assert pFA6.seguid() == 'cdseguid=FkTDjvMc5QaNMAE_dhD52VXLa6c'
```

```python
uptag_and_dntag = []
uptag_and_dns45 = []

outer_cassettes = {}

for index, row in df.iterrows():
    row_dict = row.to_dict()

    genename = row_dict["ORF_name"]

    if genename in ["YHR119W",]:
        continue

    upt = Primer(row_dict["UPTAG_primer_sequence"])
    dnt = Primer(row_dict["DNTAG_primer_sequence"])
    ups45 = Primer(row_dict["UPstream45_primer_sequence"])
    dns45 = Primer(row_dict["DNstream45_primer_sequence"])
    ups90 = Primer(row_dict["UPstream90_primer_sequence"])
    dns90 = Primer(row_dict["DNstream90_primer_sequence"])

    upt.id = f"UPTAG_{genename}"
    dnt.id = f"DNTAG_{genename}"
    dns45.id = f"DNstream45_{genename}"

    if upt and dnt:
        inner_cassette = pcr(upt, dnt, pFA6)
        uptag_and_dntag.append(genename)
        strategy = "Strategy 1: UPTAG, DNTAG"
        inner_cassette.annotations["comment"] = strategy
        try:
            outer_cassette = pcr(ups45, dns45, inner_cassette)
        except ValueError:
            dns45.seq = dns45.seq.replace("GATGTCCACGAGCTCTCT", U1)
            try:
                outer_cassette = pcr(ups45, dns45, inner_cassette)
            except ValueError:
                outer_cassette = "No outer cassette"
        
    elif upt and dns45:
        outer_cassette = pcr(upt, dns45, pFA6)
        uptag_and_dns45.append(genename)
        strategy = "Strategy 2: UPTAG, DNstream"
        outer_cassette.annotations["comment"] = strategy
    outer_cassettes[genename] =  outer_cassette
```

The gene [YML100W-A](https://www.yeastgenome.org/locus/S000028688) (174bp) in the primer list seem to correspond to a different gene on chromosome 13.

This gene is called [YML099W-A](https://www.yeastgenome.org/locus/S000004567) (330bp) located 3918 bp upstream of YML100W-A. 

The YML099W-A was present in old version(s) of the yeast genome but not in the latest.

To complicate things further, the YML099W-A (330bp) is also present in [UNIPROT](https://www.uniprot.org/uniprot/Q6B2I9). From UNIPROT, there is a link back to Genbank [AY692741.1](https://www.ncbi.nlm.nih.gov/nuccore/AY692741) where the **same** 330 bp sequence is referred to as YML100W-A. 

These contradictions might be due to mislabeling of this gene in the primer list or at some other point. 

The YML099W-A is a dubious gene and overlaps with [ARG81(YML099C)](https://www.yeastgenome.org/locus/S000004565). This means that the KanMX4 cassette in the  YML100W-A deletion strain is likely to inactivate both YML099W-A and ARG81 instead of YML100W-A.

Additionally, there is a protein record in Genbank [Q6B2I9.1](https://www.ncbi.nlm.nih.gov/protein/Q6B2I9.1) where it is claimed (incorrectly) that YML099W-A and YML100W-A are synonyms.
