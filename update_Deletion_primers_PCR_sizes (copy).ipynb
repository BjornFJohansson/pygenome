```python
import pickle
import csv
from collections import defaultdict as dd

primers = dd(list)

counter = 0

with open("Deletion_primers_PCR_sizes.txt", 'rt') as csvfile:
    rd = csv.reader(csvfile, delimiter='\t')
    keys = [k.strip() for k in next(rd)]
    next(rd)
    for row in rd:
        counter+=1
        values = [x.strip() for x in row]
        primers[values[1]].append({k:v for k,v in zip(keys,values)})
```

```python
pFA6a_kanMX4 = pickle.load((data_dir/"pFA6a-kanMX4.pickle").open("rb"))
```

```python
from pydna.primer import Primer
from pydna.amplify import pcr
```

```python
uptag_and_dntag = []
uptag_and_dns45 = []

for genename, primerlist in primers.items():
    for p in primerlist:
        upt = Primer(p["UPTAG_primer_sequence"])
        dnt = Primer(p["DNTAG_primer_sequence"])
        dns45 = Primer(p["DNstream45_primer_sequence"])

        upt.id = f"UPTAG_{genename}"
        dnt.id = f"DNTAG_{genename}"
        dns45.id = f"DNstream45_{genename}"
        
        if upt and dnt:
            inner_cassette = pcr(upt, dnt, pFA6a_kanMX4)
            p["Strategy"] = 1
            uptag_and_dntag.append(genename)
        elif upt and dns45:
            inner_cassette = pcr(upt, dns45, pFA6a_kanMX4)
            p["Strategy"] = 2
            uptag_and_dns45.append(genename)
        else:
            print(genename, "no strategy")
        p["Inner_cassette"] = inner_cassette
```

Most genes were deleted using the Strategy 1, but for about three percent of the deletions there was no DNTAG_primer_sequence entry in the original table.

Instead, the DNstream45_primer_sequence anneals directly to the pFA6a_kanMX4 sequence (Strategy 2).

### Outer cassette

```python
no_outer_cassette = []
for genename, primerlist in primers.items():
    for p in primerlist:
    
        ups45 = Primer(p["UPstream45_primer_sequence"])
        ups45.id = f"UPstream45_{genename}"

        dns45 = Primer(p["DNstream45_primer_sequence"])
        dns45.id = f"DNstream45_{genename}"
        
        inner_cassette = p["Inner_cassette"]
        
        try:
            outer_cassette = pcr(ups45, dns45, inner_cassette)
        except ValueError as ve:
            no_outer_cassette.append(genename)
```

For 354 genes, there was no product from the outer cassette amplification. These genes are on chromosomes 1, 5 and 13 (A, E and M).

All the 354 genes share a feature. The 3' part of the UPstream45_primer_sequence has the U1 sequence, but for a single nucleotide difference.


    GATGTCCACGAGGTCTCT   Canonical U1 sequence (Winzeler et al. 1999, Giaever et al. 2002, Giaever et al. 2014)
    ||||||||||||X|||||
    GATGTCCACGAGCTCTCT   3' part of UPstream45_primer_sequence for each of the 354 genes w/o outer cassette amplification.



```python
canonic_u1 = "GATGTCCACGAGGTCTCT"
mutated_u1 = "GATGTCCACGAGcTCTCT"
primer_ends_ups45 = []

for genename, primerlist in primers.items():
    for p in primerlist:
        if genename in no_outer_cassette:
            ups45 = p["UPstream45_primer_sequence"]
            primer_ends_ups45.append(ups45[-18:])
        break
```

```python
genes_with_new_ups45_primer = []

for genename, primerlist in primers.items():
    for p in primerlist:
        if genename in no_outer_cassette:
            ups45 = p["UPstream45_primer_sequence"]
            if mutated_u1.upper() in ups45: 
                p["New_UPstream45_primer_sequence"] = ups45.replace(mutated_u1.upper(), canonic_u1)
                genes_with_new_ups45_primer.append(genename)
            else:
                p["New_UPstream45_primer_sequence"] = ""
        else:
            p["New_UPstream45_primer_sequence"] = ""           
```

```python
from pydna.assembly import Assembly
from tqdm.notebook import tqdm
```

The cell below might take a long time (~ 20 min) to run.

```python
cassette_recombination = []
no_cassette_recombination = []

for genename, primerlist in tqdm(primers.items()):
    
    if genename in genes:
        gene = genes[genename]
    else:
        continue
    
    locus = gene.locus()
    
    for p in primerlist:
        
        outer_cassette = p["Outer_cassette"]
        
        asm = Assembly((locus, outer_cassette, locus))      

        candidates = asm.assemble_linear()
        
        try:
            kanmx4_gene = candidates[0]
        except:
            no_cassette_recombination.append(genename)
        else:
            cassette_recombination.append(genename)
        break
        
```

The gene [YML100W-A](https://www.yeastgenome.org/locus/S000028688) (174bp) in the primer list seem to correspond to a different gene on chromosome 13.

This gene is called [YML099W-A](https://www.yeastgenome.org/locus/S000004567) (330bp) located 3918 bp upstream of YML100W-A. 

The YML099W-A was present in old version(s) of the yeast genome but not in the latest.

To complicate things further, the YML099W-A (330bp) is also present in [UNIPROT](https://www.uniprot.org/uniprot/Q6B2I9). From UNIPROT, there is a link back to Genbank [AY692741.1](https://www.ncbi.nlm.nih.gov/nuccore/AY692741) where the **same** 330 bp sequence is referred to as YML100W-A. 

These contradictions might be due to mislabeling of this gene in the primer list or at some other point. 

The YML099W-A is a dubious gene and overlaps with [ARG81(YML099C)](https://www.yeastgenome.org/locus/S000004565). This means that the KanMX4 cassette in the  YML100W-A deletion strain is likely to inactivate both YML099W-A and ARG81 instead of YML100W-A.

Additionally, there is a protein record in Genbank [Q6B2I9.1](https://www.ncbi.nlm.nih.gov/protein/Q6B2I9.1) where it is claimed (incorrectly) that YML099W-A and YML100W-A are synonyms.
