# Pygenome example for the *TDH3* gene

```python
from pygenome.saccharomyces_cerevisiae.S288C import gene_dicts
```

```python
sysgenedict, stdgenedict = gene_dicts()
```

```python
gapdh = stdgenedict["TDH3"]
```

```python
type(gapdh)
```

```python
gapdh
```

```python
promoter = gapdh.promoter
```

```python
dna = gapdh.dna
```

```python
dna
```

```python
type(dna)
```

```python
cds = gapdh.cds
```

```python
cds
```

```python
cds.seq
```

```python
terminator = gapdh.terminator
```

```python
terminator
```

```python
type(terminator)
```

```python
gapdh.transcriptional_unit.seq == (promoter+dna+terminator).seq
```

```python
PDX1 = gapdh.succ
PDX1
```

```python
HIP1 = gapdh.pred
HIP1
```

```python
PDX1.terminator.seq == gapdh.promoter.seq
```

```python
gapdh.tandem
```

```python
HIP1.terminator.seq.reverse_complement() == gapdh.terminator.seq
```
